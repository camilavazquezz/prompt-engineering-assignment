# Exercise 1: Prompt Chaining for a Customer Support AI

Tools used:
- Google Colab
- Python
- ChatGPT for prompt design and testing

Goal:
Create a multi-step customer-support flow where the output from one
prompt becomes input for the next prompt.

```python
customer_message = """
I cannot log into my account. I reset my password twice, but the reset
link keeps saying it has expired. I have an exam tomorrow and urgently
need access.
"""

print("Customer message:")
print(customer_message)
```

## Step 1: Classify the Customer’s Issue

This step identifies the type of customer problem, explains the issue,
and determines its urgency. The classification will be used as input
for Step 2.

```python
prompt_1 = f"""
You are a customer-support intent classifier.

Classify the customer message as one of the following:
- billing
- technical support
- account access
- general question

Return the result in this format:
1. Category
2. One-sentence explanation
3. Urgency level: low, medium, or high

Customer message:
{customer_message}
"""

print("Prompt 1:")
print(prompt_1)
```

```python
classification = {
    "category": "account access",
    "explanation": "The customer cannot access their account because the password-reset link is not working.",
    "urgency": "high"
}

print("Step 1 output:")
print(classification)
```

## Step 2: Identify Missing Information

This step uses the original customer message and the classification from
Step 1. It identifies the information needed to investigate the problem.

The prompt limits the assistant to two questions and prevents it from
requesting sensitive information such as passwords.

```python
prompt_2 = f"""
You are a customer-support information-gathering assistant.

Using the customer message and the classification from Step 1, identify
the information needed to investigate the issue.

Ask for no more than two pieces of information.
Do not request the customer’s password or complete payment information.

Customer message:
{customer_message}

Classification from Step 1:
{classification}

Return the requested information as a numbered list.
"""

print("Prompt 2:")
print(prompt_2)
```

```python
missing_information = [
    "The email address connected to the account",
    "The exact error message shown after using the reset link"
]

print("Step 2 Output:")
for number, item in enumerate(missing_information, start=1):
    print(f"{number}. {item}")
```

## Step 3: Propose a Customer-Service Solution

This step uses the customer message, the classification from Step 1,
and the missing information from Step 2.

The response must be empathetic and professional. It must not promise
that the account will be restored immediately and must not request the
customer’s password.

```python
prompt_3 = f"""
You are a helpful customer-support representative.

Create a concise and empathetic response that:

- Acknowledges the customer’s problem
- Uses the account-access classification
- Requests the missing information from Step 2
- Explains the next step
- Does not request the customer’s password
- Does not guarantee immediate account access
- Uses a professional tone

Customer message:
{customer_message}

Classification from Step 1:
{classification}

Missing information from Step 2:
{missing_information}
"""

print("Prompt 3:")
print(prompt_3)
```

```python
customer_response = """
I’m sorry that you are having trouble accessing your account. Please
provide the email address connected to the account and the exact error
message shown when you use the reset link. We will use this information
to verify the issue and help restore your access. Please do not send us
your password.
"""

print("Step 3 Output:")
print(customer_response)
```

## Step 4: Decide Whether to Escalate

This step uses the issue classification, missing information, and proposed
customer response from the previous steps.

The issue should be escalated because the customer has already attempted
the password-reset process twice and the reset link continues to fail.

```python
prompt_4 = f"""
You are a customer-support escalation assistant.

Decide whether the issue should be labeled AUTOMATE or ESCALATE.

Escalate the issue if:
- The customer cannot access the account after repeated attempts
- The password-reset process continues to fail
- The issue requires identity verification by a human representative

Return:
1. AUTOMATE or ESCALATE
2. One-sentence reason

Issue classification:
{classification}

Missing information:
{missing_information}

Proposed customer response:
{customer_response}
"""

print("Prompt 4:")
print(prompt_4)
```

```python
escalation_decision = {
    "decision": "ESCALATE",
    "reason": "The customer has already attempted the password-reset process twice, but the reset link continues to fail."
}

print("Step 4 Output:")
print(escalation_decision)
```

## Prompt Chain Summary

The prompt chain follows this process:

1. Classify the customer’s issue.
2. Use the classification to identify missing information.
3. Use the classification and missing information to create a response.
4. Use the earlier information and response to decide whether escalation
   is needed.

Each step depends on information produced by the previous steps.

```python
assert classification["category"] in prompt_2
assert str(classification) in prompt_2
assert str(missing_information) in prompt_3
assert str(classification) in prompt_3
assert str(customer_response) in prompt_4
assert str(classification) in prompt_4

print("Chain validation: PASS")
print("Each prompt successfully used information from an earlier step.")
```

## Testing and Iteration

Initial prompt:

Help the customer log into their account.

Initial result:

Try resetting your password again.

Problem identified:

The response repeated a solution that the customer had already tried.
It also did not ask for the error message or protect the customer’s
password.

Improved prompt:

Using the account-access classification and missing information, write a
concise and empathetic response. Ask for the account email and exact error
message, do not request the customer’s password, and explain that the issue
will be escalated if the reset process continues to fail.

Improved result:

I’m sorry that you are having trouble accessing your account. Please
provide the email address connected to the account and the exact error
message shown when you use the reset link. We will use this information
to verify the issue and help restore your access. Please do not send us
your password.

The improved prompt was more effective because it included the customer’s
previous attempts, specific information to request, a security restriction,
and a clear next step.

```python
print("FINAL WORKFLOW TEST: PASS")
print("Customer issue classified:", classification["category"])
print("Missing information identified:", len(missing_information), "items")
print("Customer response created: Yes")
print("Escalation decision:", escalation_decision["decision"])
print("Prompt improvement demonstrated: Yes")
```
