# Exercise 3: Self-Reflection Prompt for Improving Output

Tools used:
- Google Colab
- Python
- ChatGPT for prompt design and testing

Goal:
Create an original summary, ask AI to critique it using specific
requirements, and produce an improved summary.

## Source Information

The summary will be based on the following information:

Artificial sweeteners are low- or no-calorie ingredients used to add
sweetness to foods and beverages. They can help reduce sugar and calorie
intake, but researchers continue to study their possible effects on the
gut microbiome and long-term health. Current research does not provide
one clear conclusion, and individual responses may vary. Therefore,
artificial sweeteners should not automatically be considered healthier
for every person.

```python
source_text = """
Artificial sweeteners are low- or no-calorie ingredients used to add
sweetness to foods and beverages. They can help reduce sugar and calorie
intake, but researchers continue to study their possible effects on the
gut microbiome and long-term health. Current research does not provide
one clear conclusion, and individual responses may vary. Therefore,
artificial sweeteners should not automatically be considered healthier
for every person.
"""

print("Source information:")
print(source_text)
```

## Original Summary Prompt

This prompt asks for a short summary, but it does not provide enough
constraints about accuracy, audience, word count, or formatting. This
allows weaknesses to appear in the first version.

```python
original_prompt = f"""
Summarize the following information in a few sentences:

{source_text}
"""

original_summary = """
Artificial sweeteners are healthy because they have no calories. They
do not affect the body and are better than sugar for everyone.
"""

print("Original Prompt:")
print(original_prompt)

print("\nOriginal Summary:")
print(original_summary)
```

## Self-Reflection and Revision Prompt

The prompt below asks AI to critique the original summary before rewriting
it. It includes specific criteria for accuracy, clarity, audience, length,
focus, and formatting.

```python
self_reflection_prompt = f"""
You are an editing and fact-checking assistant.

Review the original summary against the source information and complete
the following two tasks:

Task 1: Identify three specific problems with the original summary.
Check for:
- Accuracy
- Unsupported or exaggerated claims
- Clarity
- Focus on the main idea
- Appropriate language for a college-level audience

Task 2: Write a revised summary.

Requirements for the revised summary:
- Write for a college student
- Use clear and neutral language
- Use no more than 75 words
- Include the possible benefit of reducing sugar or calories
- Mention that research on long-term health effects is still developing
- Do not claim that artificial sweeteners are always healthy or unhealthy
- Do not add information that is not supported by the source
- Use one organized paragraph
- Provide only the critique and revised summary

Source information:
{source_text}

Original summary:
{original_summary}
"""

print("Self-Reflection Prompt:")
print(self_reflection_prompt)
```

```python
critique = """
1. The summary incorrectly claims that artificial sweeteners are healthy
for everyone.
2. It makes an unsupported claim that artificial sweeteners do not affect
the body.
3. It leaves out the uncertainty surrounding research on the gut
microbiome and long-term health effects.
"""

improved_summary = """
Artificial sweeteners are low- or no-calorie ingredients that can help
reduce sugar and calorie intake. However, researchers are still studying
their possible effects on the gut microbiome and long-term health.
Because individual responses may vary and research has not reached one
clear conclusion, artificial sweeteners should not automatically be
considered healthier for everyone.
"""

print("Self-Critique:")
print(critique)

print("\nImproved Summary:")
print(improved_summary)
```

## Demonstrated Improvement

The original summary was inaccurate and overly broad because it stated that
artificial sweeteners were healthy for everyone and did not affect the body.

The improved summary removes those unsupported claims. It includes the
possible benefit of reducing sugar and calories, explains that research is
still developing, and uses a neutral tone appropriate for a college student.

```python
original_word_count = len(original_summary.split())
improved_word_count = len(improved_summary.split())

print("Original summary word count:", original_word_count)
print("Improved summary word count:", improved_word_count)

print("\nOriginal and improved summaries are different:",
      original_summary.strip() != improved_summary.strip())

print("Improved summary is within the 75-word limit:",
      improved_word_count <= 75)
```

## Testing and Validation

The improved summary was checked to make sure it:

- Is different from the original summary
- Is no more than 75 words
- Uses neutral language
- Includes the potential benefit of reducing sugar and calories
- Mentions that research is still developing
- Avoids claiming that artificial sweeteners are always healthy or unhealthy

```python
assert original_summary.strip() != improved_summary.strip()
assert improved_word_count <= 75
assert "sugar and calorie intake" in improved_summary
assert "research" in improved_summary.lower()
assert "not automatically" in improved_summary.lower()

print("FINAL VALIDATION: PASS")
print("The summary was critiqued and improved according to the requirements.")
```

## Final Reflection

This exercise demonstrates self-reflection prompting because the AI first

a revised version. The revised summary is more accurate, clearer, more
neutral, and better aligned with the intended college-student audience.
