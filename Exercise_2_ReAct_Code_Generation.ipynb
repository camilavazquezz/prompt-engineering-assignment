# Exercise 2: Code Generation with ReAct Prompting

Tools used:

- Python
- ChatGPT for prompt design and code generation

Goal:
Use a ReAct-style prompt to plan, generate, test, observe, and improve
Python code.

## Task

Create a Python program that calculates the average of a list of student
grades.

The program must:
- Accept a list of grades as input
- Calculate the average
- Round the answer to two decimal places
- Handle an empty list without crashing
- Ignore grades outside the valid range of 0 to 100
- Use Python only
- Use no outside libraries

## ReAct Prompt

You are a Python programming assistant. Complete the task using the
following stages:

1. Plan:
   Briefly explain how the problem should be solved.

2. Generate:
   Write Python code that calculates the average of a list of grades.

3. Run:
   Test the code using this sample list:
   [88, 92, 79, 95, 86]

4. Observe:
   Check whether the output is correct and look for possible problems,
   including an empty list or grades outside the range of 0 to 100.

5. Fix:
   If there are errors or missing edge cases, revise the code.

6. Finalize:
   Provide the final working Python code and briefly explain the output.

Requirements:
- Use Python only.
- Do not use outside libraries.
- The program must accept a list of grades.
- Valid grades must be between 0 and 100.
- Ignore grades outside that range.
- Handle an empty list without crashing.
- Round the average to two decimal places.
- Clearly label the output.

```python
react_prompt = """
You are a Python programming assistant. Complete the task using the
following stages:

1. Plan the solution.
2. Generate the Python code.
3. Run the code using sample input.
4. Observe the output and check for errors.
5. Fix any errors or missing edge cases.
6. Provide the final working code.

Requirements:
- Use Python only.
- Do not use outside libraries.
- Accept a list of grades.
- Only grades from 0 to 100 are valid.
- Ignore invalid grades.
- Handle an empty list without crashing.
- Round the average to two decimal places.
- Clearly label the output.
"""

print(react_prompt)
```

## ReAct Planning Stage

The program will use a function that receives a list of grades. It will
remove grades outside the range of 0 to 100, check whether any valid
grades remain, calculate the average, and round the result to two decimal
places.

## ReAct Code Generation Stage

The following code was generated based on the requirements in the ReAct
prompt.

```python
def calculate_average(grades):
    valid_grades = [
        grade for grade in grades
        if isinstance(grade, (int, float)) and 0 <= grade <= 100
    ]

    if len(valid_grades) == 0:
        return "No valid grades were provided."

    average = sum(valid_grades) / len(valid_grades)
    return round(average, 2)


grades = [88, 92, 79, 95, 86]

average = calculate_average(grades)

print("Grades:", grades)
print("Average:", average)
```

## ReAct Run and Observation Stage

The program was tested using five valid grades. The expected result is
88.0 because the total of the grades divided by the number of grades is
88.

```python
expected_average = 88.0

print("Expected average:", expected_average)
print("Actual average:", average)

if average == expected_average:
    print("Test result: PASS")
else:
    print("Test result: FAIL")
```

## ReAct Fix Stage

During observation, the code was checked for two edge cases:

1. An empty list of grades
2. Grades outside the valid range of 0 to 100

The code was designed to ignore invalid grades and return a message
instead of crashing when no valid grades are available.

```python
test_cases = {
    "Normal grades": [88, 92, 79, 95, 86],
    "Grades with invalid values": [88, 105, 92, -5, 79],
    "Empty list": []
}

for test_name, test_grades in test_cases.items():
    print(test_name)
    print("Input:", test_grades)
    print("Output:", calculate_average(test_grades))
    print()
```

## ReAct Finalization

The final code successfully calculates the average of valid grades,
ignores grades outside the range of 0 to 100, and handles an empty list
without producing an error.

```python
print("FINAL CODE TEST: PASS")
print("The program handled normal input, invalid grades, and an empty list.")
```
